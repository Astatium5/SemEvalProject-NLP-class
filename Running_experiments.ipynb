```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import f1_score, confusion_matrix, classification_report
import glob # This is great for finding all your results files

# Define our label names
label_names_st1 = ["Clear Reply", "Ambivalent", "Clear Non-Reply"]
label_names_st2 = [
    "Explicit", "Implicit", "Dodging", "Deflection",
    "Partial", "General", "Declining", "Ignorance", "Clarification"
]

# --- 1. LOAD ALL EXPERIMENT RESULTS ---

# Find all the CSV files your scripts created
results_files = glob.glob("./results/*.csv")
results_data = {}

print(f"Found {len(results_files)} results files.")

for f in results_files:
    # Get a clean name like "roberta-base-direct_clarity"
    experiment_name = f.split('/')[-1].replace('.csv', '')
    print(f"Loading: {experiment_name}")
    results_data[experiment_name] = pd.read_csv(f)

# Example: Access your first DataFrame
# df_roberta_st1 = results_data['roberta-base-direct_clarity']


# --- 2. CALCULATE F1 AND PLOT CONFUSION MATRIX (for Subtask 1) ---

experiment_to_plot = 'roberta-base-direct_clarity' # Change this to see other results
df_plot = results_data[experiment_to_plot]

# Calculate Macro F1-Score
f1 = f1_score(df_plot['true_labels'], df_plot['pred_labels'], average='macro')
print(f"\n--- Results for {experiment_to_plot} ---")
print(f"Macro F1-Score: {f1:.4f}")

# Print a full classification report
print("\nClassification Report:")
print(classification_report(df_plot['true_labels'], df_plot['pred_labels'], target_names=label_names_st1))

# Plot the confusion matrix
print("Plotting Confusion Matrix...")
cm = confusion_matrix(df_plot['true_labels'], df_plot['pred_labels'], labels=label_names_st1)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=label_names_st1,
            yticklabels=label_names_st1)
plt.title(f'Confusion Matrix for {experiment_to_plot}', fontsize=16)
plt.xlabel('Predicted Label', fontsize=12)
plt.ylabel('True Label', fontsize=12)
plt.show()


# --- 3. PERFORM THE "TWO-STEP" METHOD ANALYSIS ---
# This is the key insight from the paper!

print("\n--- Performing Two-Step 'Evasion-Based Clarity' Analysis ---")

# Define the mapping from 9 labels (ST2) to 3 labels (ST1)
# !! CHECK THE PAPER FOR THE REAL MAPPING !! This is just an example
evasion_to_clarity_map = {
    'Explicit': 'Clear Non-Reply',
    'Implicit': 'Ambivalent',
    'Dodging': 'Ambivalent',
    'Deflection': 'Ambivalent',
    'Partial': 'Ambivalent',
    'General': 'Ambivalent',
    'Declining': 'Clear Non-Reply',
    'Ignorance': 'Clear Non-Reply',
    'Clarification': 'Clear Non-Reply'
}

# Load the relevant Subtask 1 and Subtask 2 results
# (Make sure you ran these experiments first!)
try:
    df_st1_direct = results_data['deberta-v3-base-direct_clarity']
    df_st2_evasion = results_data['deberta-v3-base-evasion_based_clarity']

    # 1. Get the "direct" F1 score from the ST1 experiment
    f1_direct = f1_score(df_st1_direct['true_labels'], df_st1_direct['pred_labels'], average='macro')

    # 2. Create the "two-step" predictions
    # Get the true labels from the Subtask 1 file (they are the same)
    true_labels_st1 = df_st1_direct['true_labels']

    # Get the 9-class predictions from the Subtask 2 file
    preds_st2 = df_st2_evasion['pred_labels']

    # Map the 9-class predictions down to 3 classes
    preds_two_step = [evasion_to_clarity_map[label] for label in preds_st2]

    # 3. Get the "two-step" F1 score
    f1_two_step = f1_score(true_labels_st1, preds_two_step, average='macro')

    # 4. COMPARE!
    print("\n--- Comparison for DeBERTa ---")
    print(f"  Direct Clarity (ST1) F1:     {f1_direct:.4f}")
    print(f"  Evasion-Based (ST2 -> ST1) F1: {f1_two_step:.4f}")

    if f1_two_step > f1_direct:
        print("\nSUCCESS: The two-step method is better, just as the paper predicted!")
    else:
        print("\nINTERESTING: The direct method was better. Why?")

except KeyError:
    print("\nSkipping two-step analysis.")
    print("Make sure you have run and exported results for both:")
    print("  - 'deberta-v3-base-direct_clarity'")
    print("  - 'deberta-v3-base-evasion_based_clarity'")
```
